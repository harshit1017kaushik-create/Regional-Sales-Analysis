```python
import pandas as pd

ml_data = pd.read_csv('Sales_data_Regional_Sales_Final.csv')

print(ml_data.shape)
ml_data.head()
```

```python
summary = {
    'Total Revenue': ml_data['revenue'].sum(),
    'Total Cost': ml_data['total_cost'].sum(),
    'Total Profit': ml_data['profit'].sum(),
    'Average Profit Margin (%)': ml_data['profit_margin_pct'].mean(),
    'Total Quantity Sold': ml_data['quantity'].sum(),
    'Number of Orders': ml_data['order_number'].nunique(),
    'Number of Customers': ml_data['customer_name'].nunique(),
    'Number of Products': ml_data['product_name'].nunique()
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=['Metric', 'Value']
)

summary_df
```

```python
channel_summary = (
    ml_data.groupby('channel')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        quantity=('quantity', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
    .sort_values('revenue', ascending=False)
)

channel_summary
```

```python
product_summary = (
    ml_data.groupby('product_name')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        quantity=('quantity', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
    .sort_values('revenue', ascending=False)
)

product_summary.head(10)
```

```python
region_summary = (
    ml_data.groupby('us_region')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        quantity=('quantity', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
    .sort_values('revenue', ascending=False)
)

region_summary
```

```python
customer_summary = (
    ml_data.groupby('customer_name')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        quantity=('quantity', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
    .sort_values('revenue', ascending=False)
)

customer_summary.head(10)
```

```python
monthly_summary = (
    ml_data.groupby('order_month')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        quantity=('quantity', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
    .sort_index()
)

monthly_summary.head(12)
```

```python
margin_summary = (
    ml_data.groupby('product_name')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
    .sort_values('avg_margin', ascending=False)
)

margin_summary
```

### Product Margin Insights

- Product 9 has the highest average profit margin at approximately 39.90%.
- Product 26 generates the highest revenue and total profit, although its average margin is lower than the highest-margin products.
- Product 2 has the lowest average margin at approximately 35.88%.
- Revenue performance and margin performance differ, so high sales volume does not necessarily mean the highest profitability rate.

```python
customer_margin_summary = (
    ml_data.groupby('customer_name')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
    .sort_values('avg_margin', ascending=False)
)

customer_margin_summary.head(15)
```

### Customer Margin Insights

- Revitol Company has the highest average profit margin among the customers shown, at approximately 42.85%.
- Neutrogena Ltd and Safecor Corp also show average margins above 42%.
- Organon Corp combines relatively high revenue with a strong average margin of approximately 40.47%.
- Customer profitability varies, showing that high revenue and high profit margin are different business dimensions.

```python
product_profit_analysis = (
    ml_data.groupby('product_name')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
    .sort_values('profit', ascending=False)
)

product_profit_analysis.head(10)
```

### Product Profit Insights

- Product 26 generates the highest total profit at approximately $44.41M and also has the highest revenue.
- Product 25 ranks next in total revenue and profit among the products shown.
- Product 5 and Product 1 have higher average margins than some higher-revenue products, but their total profits are lower.
- The analysis shows that total profit depends on both revenue volume and profit margin.

```python
product_opportunity = (
    ml_data.groupby('product_name')
    .agg(
        revenue=('revenue', 'sum'),
        profit=('profit', 'sum'),
        avg_margin=('profit_margin_pct', 'mean')
    )
)

product_opportunity['revenue_rank'] = (
    product_opportunity['revenue']
    .rank(ascending=False)
)

product_opportunity['margin_rank'] = (
    product_opportunity['avg_margin']
    .rank(ascending=False)
)

product_opportunity.sort_values(
    ['revenue_rank', 'margin_rank']
).head(10)
```

### Product Revenue vs Margin Insights

- Product 26 leads in revenue and total profit, while its average-margin rank is lower than its revenue rank.
- Product 5 combines relatively high revenue with a comparatively strong average margin.
- Product 1 has a lower revenue rank but a relatively higher margin rank.
- Product 2 generates substantial revenue but has a relatively low average margin.
- Revenue volume and profit margin should therefore be analyzed together when evaluating product performance.
