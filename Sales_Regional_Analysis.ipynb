```python
monthly_sales = (
    sales_orders
    .groupby('order_month', as_index=False)['Line Total']
    .sum()
)
```

```python
monthly_sales.head()
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Convert order_month to a real datetime value
monthly_sales['order_month'] = monthly_sales['order_month'].dt.to_timestamp()

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales['order_month'],
    monthly_sales['Line Total'],
    linewidth=2
)

plt.title('Monthly Revenue Trend (2022–2025)', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Revenue')

# Show only a few date labels to keep the graph clean
plt.gca().xaxis.set_major_locator(mdates.MonthLocator(interval=3))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))

plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
top_products = (
    sales_orders
    .groupby('Product Name')['Line Total']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

top_products.sort_values().plot(
    kind='barh'
)

plt.title('Top 10 Products by Revenue', fontsize=16)
plt.xlabel('Revenue')
plt.ylabel('Product')

plt.tight_layout()
plt.show()
```

```python
product_profit = (
    sales_orders
    .groupby('Product Name')['profit']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

product_profit
```

```python
numeric_data = sales_orders[
    [
        'Order Quantity',
        'Unit Price',
        'Line Total',
        'Total Unit Cost',
        'total_cost',
        'profit',
        'profit_margin_pct'
    ]
]

correlation_matrix = numeric_data.corr()

correlation_matrix
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap', fontsize=16)
plt.tight_layout()
plt.show()
```
