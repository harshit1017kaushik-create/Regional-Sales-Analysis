```python
ml_data = ml_check.drop_duplicates().reset_index(drop=True)
```

```python
print("Shape:", ml_data.shape)
print("Duplicates:", ml_data.duplicated().sum())
print("Missing values:", ml_data.isnull().sum().sum())
```

```python
ml_data.describe(include='all').T
```

```python
ml_data.nunique().sort_values()
```

```python
ml_data['margin_category'] = ml_data['profit_margin_pct'].apply(
    lambda x: 'High Margin' if x >= 40 else 'Low Margin'
)

print(ml_data['margin_category'].value_counts())
print()
print(ml_data['margin_category'].value_counts(normalize=True) * 100)
```

```python
features = [
    'quantity',
    'unit_price',
    'unit_cost',
    'order_month_num',
    'channel',
    'warehouse_code',
    'customer_name',
    'product_name',
    'state_code',
    'us_region'
]

X = ml_data[features]
y = ml_data['margin_category']

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
numerical_features = [
    'quantity',
    'unit_price',
    'unit_cost',
    'order_month_num'
]
```

```python
categorical_features = [
    'channel',
    'warehouse_code',
    'customer_name',
    'product_name',
    'state_code',
    'us_region'
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numerical_features = [
    'quantity',
    'unit_price',
    'unit_cost',
    'order_month_num'
]

categorical_features = [
    'channel',
    'warehouse_code',
    'customer_name',
    'product_name',
    'state_code',
    'us_region'
]

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

print("Preprocessor created successfully.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', LogisticRegression(max_iter=1000))
    ]
)

print("Logistic Regression pipeline created successfully.")
```

```python
logistic_pipeline.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
```

```python
y_pred = logistic_pipeline.predict(X_test)

print("Predictions generated successfully.")
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(
    y_test,
    y_pred,
    pos_label='High Margin'
)
recall = recall_score(
    y_test,
    y_pred,
    pos_label='High Margin'
)
f1 = f1_score(
    y_test,
    y_pred,
    pos_label='High Margin'
)

print("Logistic Regression Results")
print("-" * 35)
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=['Low Margin', 'High Margin']
)

print("Confusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Low Margin', 'High Margin']
)

disp.plot()
plt.title('Logistic Regression - Confusion Matrix')
plt.show()
```

```python
# Features without direct price/cost information

features_no_price_cost = [
    'quantity',
    'order_month_num',
    'channel',
    'warehouse_code',
    'customer_name',
    'product_name',
    'state_code',
    'us_region'
]

X2 = ml_data[features_no_price_cost]
y2 = ml_data['margin_category']

print("X2 shape:", X2.shape)
print("Features:")
print(X2.columns.tolist())
```

```python
X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2,
    y2,
    test_size=0.20,
    random_state=42,
    stratify=y2
)

print("X2_train:", X2_train.shape)
print("X2_test :", X2_test.shape)
print("y2_train:", y2_train.shape)
print("y2_test :", y2_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numerical_features_2 = [
    'quantity',
    'order_month_num'
]

categorical_features_2 = [
    'channel',
    'warehouse_code',
    'customer_name',
    'product_name',
    'state_code',
    'us_region'
]

preprocessor_2 = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features_2),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features_2)
    ]
)

print("Second preprocessor created successfully.")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_pipeline_2 = Pipeline(
    steps=[
        ('preprocessor', preprocessor_2),
        ('model', LogisticRegression(max_iter=1000))
    ]
)

print("Second Logistic Regression pipeline created successfully.")
```

```python
logistic_pipeline_2.fit(X2_train, y2_train)

print("Second Logistic Regression model trained successfully.")
```

```python
y2_pred = logistic_pipeline_2.predict(X2_test)

print("Predictions generated successfully.")
print("Number of predictions:", len(y2_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

accuracy_2 = accuracy_score(y2_test, y2_pred)

precision_2 = precision_score(
    y2_test,
    y2_pred,
    pos_label='High Margin'
)

recall_2 = recall_score(
    y2_test,
    y2_pred,
    pos_label='High Margin'
)

f1_2 = f1_score(
    y2_test,
    y2_pred,
    pos_label='High Margin'
)

print("Logistic Regression — Model 2")
print("-" * 40)
print(f"Accuracy : {accuracy_2:.4f}")
print(f"Precision: {precision_2:.4f}")
print(f"Recall   : {recall_2:.4f}")
print(f"F1 Score : {f1_2:.4f}")

print("\nClassification Report:")
print(classification_report(y2_test, y2_pred))
```

#### Model 1: Included price and cost variables, resulting in very high predictive performance because these variables are directly related to the calculation of profit margin.

#### Model 2: Excluded price and cost variables to test whether contextual sales information could independently predict margin category. Performance decreased substantially, indicating that contextual variables alone were much less predictive.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm_2 = confusion_matrix(
    y2_test,
    y2_pred,
    labels=['Low Margin', 'High Margin']
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_2,
    display_labels=['Low Margin', 'High Margin']
)

disp.plot()

plt.title('Logistic Regression - Model 2 Confusion Matrix')
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier

decision_tree_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor_2),
        ('model', DecisionTreeClassifier(
            max_depth=5,
            random_state=42
        ))
    ]
)

decision_tree_pipeline.fit(X2_train, y2_train)

y2_tree_pred = decision_tree_pipeline.predict(X2_test)

print("Decision Tree model trained successfully.")
print("Number of predictions:", len(y2_tree_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

accuracy_tree = accuracy_score(y2_test, y2_tree_pred)

precision_tree = precision_score(
    y2_test,
    y2_tree_pred,
    pos_label='High Margin'
)

recall_tree = recall_score(
    y2_test,
    y2_tree_pred,
    pos_label='High Margin'
)

f1_tree = f1_score(
    y2_test,
    y2_tree_pred,
    pos_label='High Margin'
)

print("Decision Tree — Model 2 Features")
print("-" * 40)
print(f"Accuracy : {accuracy_tree:.4f}")
print(f"Precision: {precision_tree:.4f}")
print(f"Recall   : {recall_tree:.4f}")
print(f"F1 Score : {f1_tree:.4f}")

print("\nClassification Report:")
print(classification_report(y2_test, y2_tree_pred))
```
